```python
def clean_shipments(raw, medians=None):
    raw = raw.drop_duplicates()
    raw['service_type'] = raw['service_type'].str.strip().str.title()
    raw['payment_mode'] = raw['payment_mode'].fillna('UnKnown')
    raw['declared_value_inr'] = pd.to_numeric(
        raw['declared_value']
        .str.replace('₹', '', regex=False)
        .str.replace('\u20b9', '', regex=False)
        .str.replace(',', '', regex=False)
        .str.strip()
        )
    raw = raw.drop(columns=['declared_value'])
    raw['dispatch_time'] = pd.to_datetime(raw['dispatch_time'])

    raw.loc[raw['parcel_weight_kg'] > 100, 'parcel_weight_kg'] = np.nan

    if medians is None:
        medians = {'parcel_weight_kg': float(raw['parcel_weight_kg'].median()),
                   'distance_km': float(raw['distance_km'].median()),
                   'declared_value_inr': float(raw['distance_km'].median())}

    raw['parcel_weight_kg'] = raw['parcel_weight_kg'].fillna(medians.get('parcel_weight_kg'))
    raw['distance_km'] = raw['distance_km'].fillna(medians.get('distance_km'))
    raw['declared_value_inr'] = raw['declared_value_inr'].fillna(medians.get('declared_value_inr'))

    return (raw, medians)
```

```python
def regression_metrics(y_true, y_pred):
    mae = mean_absolute_error(y_true, y_pred)
    mse = mean_squared_error(y_true, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_true, y_pred)

    return {
        "MAE": mae,
        "MSE": mse,
        "RMSE": rmse,
        "R2": r2
    }
```

```python
import numpy as np
import pandas as pd
from src.shipment_prep import FEATURES, NOISE_COLUMNS
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    f1_score,
    precision_score,
    recall_score,
)
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import GridSearchCV, RandomizedSearchCV
```

```python
def classification_metrics(y_true, y_pred):
    # Assumes 0 = on time and 1 = delayed.

    tn, fp, fn, tp = confusion_matrix(
        y_true,
        y_pred,
        labels=[0, 1]
    ).ravel()

    return {
        "TN": int(tn),
        "FP": int(fp),
        "FN": int(fn),
        "TP": int(tp),
        "accuracy": float(accuracy_score(y_true, y_pred)),
        "precision": float(precision_score(y_true, y_pred, zero_division=0)),
        "recall": float(recall_score(y_true, y_pred, zero_division=0)),
        "f1": float(f1_score(y_true, y_pred, zero_division=0))
    }
```

```python
tree_depths = [2, 3, 4, 5, 6, 8, 10]

best_tree_model = None
best_tree_depth = None
best_tree_f1 = -1.0

for depth in tree_depths:
    tree_model = DecisionTreeClassifier(
        criterion="gini",
        max_depth=depth,
        random_state=42
    )

    tree_model.fit(x_train, y_train)

    tree_predictions = tree_model.predict(x_val)

    current_f1 = float(
        f1_score(y_val, tree_predictions, zero_division=0)
    )

    # The list is ordered from smaller to larger depth.
    # Using > rather than >= preserves the smaller depth on an F1 tie.
    if current_f1 > best_tree_f1:
        best_tree_f1 = current_f1
        best_tree_depth = depth
        best_tree_model = tree_model

best_tree_predictions = best_tree_model.predict(x_val)

best_tree_result = classification_metrics(
    y_true=y_val,
    y_pred=best_tree_predictions
)

model_rows.append({
    "model": f"decision_tree_depth_{best_tree_depth}",
    **best_tree_result
})
```

```python
from sklearn.feature_selection import SelectKBest, f_classif

rf_all_features = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

rf_all_features.fit(x_train, y_train)

rf_importances = rf_all_features.feature_importances_

selector_all = SelectKBest(
    score_func=f_classif,
    k="all"
)

selector_all.fit(x_train, y_train)

f_scores = selector_all.scores_

feature_scores_df = pd.DataFrame({
    "feature": FEATURES,
    "f_score": [float(score) for score in f_scores],
    "rf_importance": [float(importance) for importance in rf_importances]
})

feature_scores_df["f_rank"] = (
    feature_scores_df["f_score"]
    .rank(ascending=False, method="first")
    .astype(int)
)

feature_scores_df["is_noise"] = (
    feature_scores_df["feature"].isin(NOISE_COLUMNS)
)

feature_scores_df = feature_scores_df[
    [
        "feature",
        "f_score",
        "f_rank",
        "rf_importance",
        "is_noise"
    ]
]

feature_scores_df.head(20)
```

```python
k_values = [5, 7, 10, 12, 17]

selectk_rows = []

for k in k_values:
    # A pipeline ensures the selector is fitted only on training rows.
    selectk_pipeline = Pipeline([
        ("selector", SelectKBest(f_classif, k=k)),
        ("model", RandomForestClassifier(
            n_estimators=100,
            random_state=42
        ))
    ])

    # Fit selector and forest only on training data.
    selectk_pipeline.fit(x_train, y_train)

    # Generate default-threshold (0.5) predictions on validation data.
    val_predictions = selectk_pipeline.predict(x_val)

    # Record validation F1 score.
    val_f1 = float(
        f1_score(y_val, val_predictions, zero_division=0)
    )

    selectk_rows.append({
        "k": k,
        "val_f1": val_f1
    })

selectk_df = pd.DataFrame(
    selectk_rows,
    columns=["k", "val_f1"]
)

print("\nSelectKBest validation experiment:")
print(selectk_df.round(4))
```

```python
def across_within_ratio(scaled_matrix, labels):
    """
    Calculate mean distance across clusters divided by mean distance
    within clusters, using all unordered hub pairs.
    """
    within_distances = []
    across_distances = []

    n_hubs = len(scaled_matrix)

    for i in range(n_hubs):
        for j in range(i + 1, n_hubs):
            distance = np.linalg.norm(
                scaled_matrix[i] - scaled_matrix[j]
            )

            if labels[i] == labels[j]:
                within_distances.append(distance)
            else:
                across_distances.append(distance)

    return (
        float(np.mean(across_distances))
        / float(np.mean(within_distances))
    )
```

```python
k_values = [2, 3, 4, 5]
separation_rows = []

for k in k_values:
    # Same scaled matrix, same hub order, same seed and n_init
    # for every requested K.
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = kmeans.fit_predict(X_hubs_scaled)

    # Calculate separation on scaled feature space.
    ratio = across_within_ratio(
        scaled_matrix=X_hubs_scaled,
        labels=labels
    )

    # Count hubs in each cluster and format descending sizes.
    cluster_sizes = (
        pd.Series(labels)
        .value_counts()
        .sort_values(ascending=False)
        .tolist()
    )

    cluster_sizes_text = "-".join(
        str(size)
        for size in cluster_sizes
    )

    separation_rows.append({
        "K": k,
        "across_over_within": ratio,
        "cluster_sizes": cluster_sizes_text
    })


# Required separation-evidence table.
separation_df = pd.DataFrame(
    separation_rows,
    columns=[
        "K",
        "across_over_within",
        "cluster_sizes"
    ]
)

separation_df.head()
```
