```python
def clean_shipments(raw, medians=None):
    raw = raw.drop_duplicates()
    raw['service_type'] = raw['service_type'].str.strip().str.title()
    raw['payment_mode'] = raw['payment_mode'].fillna('UnKnown')
    raw['declared_value_inr'] = pd.to_numeric(
        raw['declared_value']
        .str.replace('₹', '', regex=False)
        .str.replace('\u20b9', '', regex=False)
        .str.replace(',', '', regex=False)
        .str.strip()
        )
    raw = raw.drop(columns=['declared_value'])
    raw['dispatch_time'] = pd.to_datetime(raw['dispatch_time'])

    raw.loc[raw['parcel_weight_kg'] > 100, 'parcel_weight_kg'] = np.nan

    if medians is None:
        medians = {'parcel_weight_kg': float(raw['parcel_weight_kg'].median()),
                   'distance_km': float(raw['distance_km'].median()),
                   'declared_value_inr': float(raw['distance_km'].median())}

    raw['parcel_weight_kg'] = raw['parcel_weight_kg'].fillna(medians.get('parcel_weight_kg'))
    raw['distance_km'] = raw['distance_km'].fillna(medians.get('distance_km'))
    raw['declared_value_inr'] = raw['declared_value_inr'].fillna(medians.get('declared_value_inr'))

    return (raw, medians)
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from src.shipment_prep import FEATURES
from sklearn.preprocessing import StandardScaler
import numpy as np  


data = pd.read_csv("data/modelling_table.csv")
x, y = data[FEATURES], data["delivery_hours"]

x_train, x_temp, y_train, y_temp = train_test_split(x, y, test_size=0.3, random_state=42)
x_val, x_test, y_val, y_test = train_test_split(x_temp, y_temp, test_size=0.5, random_state=42)


print(x_train.shape, y_train.shape)
print(x_val.shape, y_val.shape)
print(x_test.shape, y_test.shape)




```

```python
scaler = StandardScaler()
scaler = scaler.fit(x_train)
print(scaler)
```

```python
x_train_scaled = scaler.transform(x_train)
x_test_scaled = scaler.transform(x_test)

train_scaled_max_abs_mean = float(np.abs(x_train_scaled.mean(axis = 0)).max())
test_scaled_max_abs_mean = float(np.abs(x_test_scaled.mean(axis = 0)).max())

print(train_scaled_max_abs_mean)
print(test_scaled_max_abs_mean)
```

```python
checks = pd.DataFrame({
    "metric": ['train_rows', 'val_rows', 'test_rows', 'train_scaled_max_abs_mean', 'test_scaled_max_abs_mean'],
    "value": [len(x_train), len(x_val), len(x_test), round(train_scaled_max_abs_mean, 4), round(test_scaled_max_abs_mean, 4)]
})


print(checks)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

def scores(y_true, pred):
    mse = mean_squared_error(y_true, pred)
    return {"MAE": mean_absolute_error(y_true, pred), "MSE": mse, "RMSE": float(np.sqrt(mse)), "R2": r2_score(y_true, pred)}


rows = [{"model": "mean_baseline",
          "split": "val", **scores(y_val, np.full(len(y_val), y_train.mean()))}]

print(rows)
```

```python
train_mean = float(y_train.mean())
baseline_pred = np.full(len(y_val), train_mean)

mae = mean_absolute_error(y_val, baseline_pred)
rmse = np.sqrt(mean_squared_error(y_val, baseline_pred))
r2 = r2_score(y_val, baseline_pred)

baseline_metrics = pd.DataFrame(
    {
        "metric": ["train_mean", "val_mae", "val_rmse", "val_r2"],
        "value": [train_mean, mae, rmse, r2],
    }
)

baseline_metrics.to_csv("outputs/part_b_baseline_checks.csv", index=False)
print(baseline_metrics.to_string(index=False))
```
